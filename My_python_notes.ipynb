```python
#grade attained and target grade
attained_grade = 60
target_grade = 50

#Comparing pf the grade

if attained_grade >= target_grade:
    print("target grade achieved")
```

```python
#grade attained and target grade
attained_grade = 60
target_grade = 50

#Comparing pf the grade

if attained_grade >= target_grade:
    print("target grade achieved")
#check if they havent hit the target
elif attained_grade < target_grade:
    print("target grade not achieved")
#check if the grade is equal to 0
elif attained_grade == 0:
    print("you have grade of 0, please contact your lecturer for assistance")
else:
    print("you have a negative grade, please contact your lecturer for assistance")
```

```python
# Check there is a key for a flight to CPT KQ792
if "KQ792" in flight_numbers:
    print("Kenya Airways has a flighT to cpt with a flight number KQ792")

else:
    #Add the flight to dictionary and then print
    flight_numbers["KQ792"] = "CPT"
    print(f"added to flight numbers: {flight_numbers}")

```

```python
#Check if there is a key for a flight to CPT KQ792
if "KQ792" and "KQ2" in flight_numbers:
    print("Kenya Airways has a flighT to cpt and New york")

else:
    #Add flight to dictionary and then print
    flight_numbers["KQ2"] = "JFK"
    #PRINT THE VALUE WHOSE KEY IS KQ2
    print (flight_numbers["KQ2"])
```
